```python

import torch
from torch import nn

#创建一个代表 CPU 的设备对象。后续把张量（Tensor）或模型放到这个设备上就会使用 CPU 运算。
torch.device('cpu')
#创建一个代表 第 0 号 GPU 的设备对象。默认为 0
torch.cuda.device('cuda')
#创建一个代表 第 1 号 GPU（即服务器或电脑上的第二张独立显卡）的设备对象。
torch.cuda.device('cuda:1')
```

```python
#查询可用gpu数量
torch.cuda.device_count()
```

```python
#查询张量所在设备,默认在 cpu
x = torch.tensor([1, 2, 3])
x.device
```

```python
#在gpu上存储张量
X = torch.ones(2,3,device=torch.device('cuda:0'))
X
```

```python
#第二个gpu上创建一个随机张量
Y = torch.rand(2,3,device=torch.device('cuda:1'))
Y
```

```python
#计算X+Y，要决定在哪个gpu上进行
Z = X.cuda(1)
print(X)
print(Z)
```

```python
Y + Z
```

```python
#神经网络
#定义一个简单的单层全连接神经网络，输入维度为 3，输出维度为 1。
net = nn.Sequential(nn.Linear(3, 1))
#将整个神经网络的权重（Weights）和偏置（Biases）等参数移动到指定的 GPU 上
net = net.to(device=torch.device('cuda:0'))


#将输入数据 X 传入网络进行前向传播计算
#grad_fn=<AddmmBackward>：表示该张量记录了计算图，后续可以用于梯度反向传播计算
net(X)
```

```python
#确认模型参数的存储设备
#net[0]：提取网络中的第一层;.weight.data：获取该层的权重矩阵数据
net[0].weight.data.device
```
